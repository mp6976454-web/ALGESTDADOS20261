```python
#16 - Crie uma matriz 4x4 com números fornecidos pelo usuário.
#Encontre o menor valor de cada coluna.

matriz = []

for linha in range(4):
    nova_linha = []

    for coluna in range(4):
        valor = int(input(f'Digite o valor [{linha}][{coluna}]: '))
        nova_linha.append(valor)

    matriz.append(nova_linha)

menores = []

for coluna in range(4):
    menor = matriz[0][coluna]

    for linha in range(4):
        if matriz[linha][coluna] < menor:
            menor = matriz[linha][coluna]

    menores.append(menor)

print('\nMatriz:')
for linha in matriz:
    print(linha)

print(f'\nMenores valores das colunas: {menores}')
```

```python
#17 - Calcule o determinante de uma matriz 3x3.

matriz = []

for linha in range(3):
    nova_linha = []

    for coluna in range(3):
        valor = float(input(f'Digite o valor [{linha}][{coluna}]: '))
        nova_linha.append(valor)

    matriz.append(nova_linha)

det = (
    matriz[0][0] * matriz[1][1] * matriz[2][2]
    + matriz[0][1] * matriz[1][2] * matriz[2][0]
    + matriz[0][2] * matriz[1][0] * matriz[2][1]
    - matriz[0][2] * matriz[1][1] * matriz[2][0]
    - matriz[0][0] * matriz[1][2] * matriz[2][1]
    - matriz[0][1] * matriz[1][0] * matriz[2][2]
)

print('\nMatriz:')
for linha in matriz:
    print(linha)

print(f'\nDeterminante = {det}')
```

```python
#18 - Leia duas matrizes 3x3 e calcule a soma e a diferença.

matriz1 = []
matriz2 = []

print('Matriz 1')

for linha in range(3):
    nova_linha = []

    for coluna in range(3):
        valor = int(input(f'Valor [{linha}][{coluna}]: '))
        nova_linha.append(valor)

    matriz1.append(nova_linha)

print('\nMatriz 2')

for linha in range(3):
    nova_linha = []

    for coluna in range(3):
        valor = int(input(f'Valor [{linha}][{coluna}]: '))
        nova_linha.append(valor)

    matriz2.append(nova_linha)

soma = []
diferenca = []

for linha in range(3):
    linha_soma = []
    linha_diferenca = []

    for coluna in range(3):
        linha_soma.append(matriz1[linha][coluna] + matriz2[linha][coluna])
        linha_diferenca.append(matriz1[linha][coluna] - matriz2[linha][coluna])

    soma.append(linha_soma)
    diferenca.append(linha_diferenca)

print('\nMatriz Soma:')
for linha in soma:
    print(linha)

print('\nMatriz Diferença:')
for linha in diferenca:
    print(linha)
```

```python
#19 - Mostre interseção, união e diferenças.

conjuntoA = set()
conjuntoB = set()

print('Conjunto A')

for i in range(10):
    conjuntoA.add(input('Digite um elemento: '))

print('\nConjunto B')

for i in range(10):
    conjuntoB.add(input('Digite um elemento: '))

print('\nInterseção:')
print(conjuntoA.intersection(conjuntoB))

print('\nUnião:')
print(conjuntoA.union(conjuntoB))

print('\nA - B:')
print(conjuntoA.difference(conjuntoB))

print('\nB - A:')
print(conjuntoB.difference(conjuntoA))
```

```python
#23 - Verifique se um conjunto é subconjunto do outro.

A = set(map(int, input('Digite os números do conjunto A: ').split()))
B = set(map(int, input('Digite os números do conjunto B: ').split()))

if A.issubset(B):
    print('A é subconjunto de B')

elif B.issubset(A):
    print('B é subconjunto de A')

else:
    print('Nenhum conjunto é subconjunto do outro')
```

```python
#25 - Reajuste salarial.

funcionarios = {}

for i in range(5):
    nome = input('Nome: ')
    salario = float(input('Salário: '))

    funcionarios[nome] = salario

for nome in funcionarios:
    if funcionarios[nome] < 2000:
        funcionarios[nome] = funcionarios[nome] * 1.10

print('\nFuncionários atualizados:')

for nome, salario in funcionarios.items():
    print(f'{nome} - R$ {salario:.2f}')
```

```python
#26 - Nome e média dos alunos.

alunos = {}

for i in range(5):
    nome = input('Nome do aluno: ')

    nota1 = float(input('Nota 1: '))
    nota2 = float(input('Nota 2: '))
    nota3 = float(input('Nota 3: '))

    alunos[nome] = [nota1, nota2, nota3]

print()

for nome, notas in alunos.items():
    media = sum(notas) / len(notas)

    print(f'Aluno: {nome} - Média: {media:.2f}')
```

```python
#29 - Produto com maior variação de preço.

produtos = {}

for i in range(3):
    nome = input('Produto: ')

    preco1 = float(input('Preço loja 1: '))
    preco2 = float(input('Preço loja 2: '))
    preco3 = float(input('Preço loja 3: '))

    produtos[nome] = [preco1, preco2, preco3]

maior_variacao = 0
produto_maior = ''

for nome, precos in produtos.items():
    variacao = max(precos) - min(precos)

    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_maior = nome

print(f'\nProduto com maior variação: {produto_maior}')
print(f'Variação: R$ {maior_variacao:.2f}')
```

```python
#30 - Contagem de palavras.

frase = input('Digite uma frase: ').lower()

palavras = frase.split()

ocorrencias = {}

for palavra in palavras:
    if palavra in ocorrencias:
        ocorrencias[palavra] += 1

    else:
        ocorrencias[palavra] = 1

print()

for palavra, quantidade in ocorrencias.items():
    print(f'{palavra}: {quantidade}')
```
